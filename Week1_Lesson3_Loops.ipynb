```python
import math
bolt_diameter_mm = 4
bolt_load_N =  4250
allowable_shear_stress_MPa = 300

shear_area_bolt_mm2 = math.pi * (bolt_diameter_mm / 2) ** 2
shear_stress_bolt_MPa = bolt_load_N / shear_area_bolt_mm2
margin_of_safety = (allowable_shear_stress_MPa / shear_stress_bolt_MPa) - 1
margin_of_safety =float("nan")
print(f"Shear Stress in MPa {shear_stress_bolt_MPa: .2f}")
print(f"Margin of Safety {margin_of_safety: .2f}")
if margin_of_safety >= 0:
    print("Bolt wil not fail")
if margin_of_safety < 0:
    print("Bolt will fail")
```
